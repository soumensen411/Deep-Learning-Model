```python
import tensorflow
from mtcnn import MTCNN
import cv2
import numpy as np
from PIL import Image
```

```python
img_pil = Image.open('/content/drive/MyDrive/Colab Notebooks/Deep Learning/CNN Project/Face Recognizer MTCNN/images/Virat+Kohli.jpg')
img = np.array(img_pil)
```

```python
import matplotlib.pyplot as plt
plt.imshow(img)
```

```python
import lz4
detector = MTCNN()
```

```python
output = detector.detect_faces(img)
```

```python
print(output)
```

```python
from google.colab.patches import cv2_imshow

x,y,width,height = output[0]['box']
img_bgr = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)
cv2.rectangle(img_bgr,pt1=(x,y),pt2=(x+width,y+height),color = (255,0,0),thickness=3)
cv2_imshow(img_bgr)
```

```python
left_eyeX,left_eyeY = output[0]['keypoints']['left_eye']
right_eyeX,right_eyeY = output[0]['keypoints']['right_eye']
noseX,noseY = output[0]['keypoints']['nose']
mouth_left_x,mouth_left_y = output[0]['keypoints']['mouth_left']
mouth_right_x,mouth_right_y = output[0]['keypoints']['mouth_right']
cv2.circle(img,center=(left_eyeX,left_eyeY),color=(0,255,0),thickness=3,radius = 5)
cv2.circle(img,center=(right_eyeX,right_eyeY),color=(0,255,0),thickness=3,radius = 5)
cv2.circle(img,center=(noseX,noseY),color=(0,255,0),thickness=3,radius = 5)
cv2.circle(img,center=(mouth_left_x,mouth_left_y),color=(0,255,0),thickness=3,radius = 5)
cv2.circle(img,center=(mouth_right_x,mouth_right_y),color=(0,255,0),thickness=3,radius = 5)



x,y,width,height = output[0]['box']
img_bgr = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)
cv2.rectangle(img_bgr,pt1=(x,y),pt2=(x+width,y+height),color = (255,0,0),thickness=3)
cv2_imshow(img_bgr)
```

```python
img2 = Image.open('/content/drive/MyDrive/Colab Notebooks/Deep Learning/CNN Project/Face Recognizer MTCNN/images/istockphoto-1368965646-612x612.jpg')
img2 = np.array(img2)
```

```python
output = detector.detect_faces(img2)
```

```python
img2 = cv2.cvtColor(img2, cv2.COLOR_RGB2BGR)
for i in output:
  x,y,width,height = i['box']
  left_eyeX,left_eyeY = i['keypoints']['left_eye']
  right_eyeX,right_eyeY = i['keypoints']['right_eye']
  noseX,noseY = i['keypoints']['nose']
  mouth_left_x,mouth_left_y = i['keypoints']['mouth_left']
  mouth_right_x,mouth_right_y = i['keypoints']['mouth_right']
  cv2.circle(img2,center=(left_eyeX,left_eyeY),color=(0,255,0),thickness=3,radius = 5)
  cv2.circle(img2,center=(right_eyeX,right_eyeY),color=(0,255,0),thickness=3,radius = 5)
  cv2.circle(img2,center=(noseX,noseY),color=(0,255,0),thickness=3,radius = 5)
  cv2.circle(img2,center=(mouth_left_x,mouth_left_y),color=(0,255,0),thickness=3,radius = 5)
  cv2.circle(img2,center=(mouth_right_x,mouth_right_y),color=(0,255,0),thickness=3,radius = 5)
  cv2.rectangle(img2,pt1=(x,y),pt2=(x+width,y+height),color = (255,0,0),thickness=3)

cv2_imshow(img2)
```
